Use AI to help you write a program that:
- Allows users to input their monthly income

- Shows a warning if budget is negative

### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE
# Print program title
print("=== Simple Budget Tracker ===")

# Ask the user to enter their monthly income
monthly_income = float(input("Enter your monthly income: $"))

# Ask the user to enter up to 3 expenses
expense1 = float(input("Enter expense 1: $"))  # first expense
expense2 = float(input("Enter expense 2: $"))  # second expense
expense3 = float(input("Enter expense 3: $"))  # third expense

# Calculate the total expenses by adding all expenses
total_expenses = expense1 + expense2 + expense3

# Calculate the remaining budget
remaining_budget = monthly_income - total_expenses

# Display the total expenses
print("\nTotal expenses: $", total_expenses)

# Display the remaining budget
print("Remaining budget: $", remaining_budget)

# Check if the remaining budget is negative
if remaining_budget < 0:
    # Show warning if user spent more than their income
    print("⚠️ Warning: Your budget is negative! You are spending more than your income.")
else:
    # Show message if the budget is still positive
    print("✅ Good job! You are within your budget.")
```
