## Part 2: Financial Decision Logic (15 minutes)

### Budget Limit Checker
Apply your if/else skills to real budget management

````python
print("=== Budget Limit Checker ===")

# Get user's budget and expense
monthly_budget = float(input("Enter your monthly budget: $"))
expense = float(input("Enter expense amount: $"))

#formula to calculate percentage of budget
percentage_of_budget=(expense/monthly_budget)*100

# Decision logic here - you write this part!
# If expense > 50% of budget: "Major purchase - think carefully"
# If expense > 25% of budget: "Significant expense"
# If expense > 10% of budget: "Moderate expense"
# Otherwise: "Minor expense - within budget"
if percentage_of_budget >= 50:
  category="Major purchase - think carefully"
elif percentage_of_budget >= 25:
  category="Significant expense"
elif percentage_of_budget >= 10:
  category="Moderate expense"
else:
  category="Minor expense - within budget"
print(f"This is {category}")
````

### Savings Goal Progress
Use multiple conditions to track savings progress:

````python
print("=== Savings Goal Tracker ===")

goal_amount = float(input("Savings goal: $"))
current_savings = float(input("Current savings: $"))

# Your task: Calculate percentage and give encouraging feedback
percentage_current_savings=(current_savings/goal_amount)*100
# 100%+: "Goal achieved! Consider a new goal."
# 75-99%: "Almost there! You're doing great."
# 50-74%: "Halfway there - keep it up!"
# 25-49%: "Good progress, stay consistent"
# Under 25%: "Just getting started - you've got this!"
if percentage_current_savings >= 100:
  category="Goal achieved! Consider a new goal"
elif percentage_current_savings >= 75:
  category="Almost there You're doing great"
elif percentage_current_savings >= 50:
  category="Halfway there - keep it up"
elif percentage_current_savings >= 25:
  category="Good progress, stay consistent"
else:
  category="Just getting started - you've got this!"
print(category)
````

## Part 3: Smart Purchase Advisor (20 minutes)

### The Challenge
Build a program that helps users make smart purchase decisions by considering multiple factors.

### Requirements

Your purchase advisor should:

1. Get purchase amount and user's available money
2. Check if they can afford it (basic affordability)
3. Give different advice based on purchase size
4. Consider their financial situation

### Starter Framework

````python
print("=== Smart Purchase Advisor ===")

# Get financial information
available_money = float(input("How much money do you have available? $"))
purchase_price = float(input("How much does the item cost? $"))
monthly_income = float(input("What's your monthly income? $"))
emergency_fund = float(input("How much is in your emergency fund? $"))
category = input("Is this a 'necessity' or 'luxury'? ").lower()

# Affordability check
if purchase_price > available_money:
    need_more = purchase_price - available_money
    print("Save up first - you need $", round(need_more,2), "more")
else:
    percent = (purchase_price / available_money) * 100
    if percent > 50:
        print("You can afford it, but it's", round(percent,1), "% of your available money. Consider waiting.")
    elif percent > 20:
        print("This seems like a reasonable purchase for your budget")
    else:
        print("This is a minor expense - go for it!")

# Emergency fund check
if available_money - purchase_price < emergency_fund:
    print("Warning: This purchase would reduce your emergency savings.")

# Category-based advice
if category == "luxury" and purchase_price > available_money * 0.3:
    print("Since this is a luxury item, you might want to wait.")
elif category == "necessity":
    print("Since this is a necessity, the purchase may be justified.")

# Payment plan calculator
if purchase_price > monthly_income:
    months = int(input("Enter number of months for payment plan: "))
    monthly_payment = purchase_price / months
    print("Monthly payment would be $", round(monthly_payment,2), "for", months, "months.")
````

### Sample Decision Logic to Implement

- **Can't afford**: "Save up first - you need $X more"
- **Can afford but expensive**: "You can afford it, but it's Y% of your available money. Consider waiting."
- **Reasonable purchase**: "This seems like a reasonable purchase for your budget"
- **Small purchase**: "This is a minor expense - go for it!"

````python
print("=== Personal Finance Tracker v0.1 ===")
print("Week 3: Smart Budget Decisions")
print()

# User setup
name = input("Enter your name: ")
monthly_income = float(input("Enter your monthly income: $"))

print(f"\nWelcome to your finance tracker, {name}!")
print(f"Monthly income: ${monthly_income:.2f}")

# Week 3 feature: Expense categorisation
print("\n--- Expense Entry ---")
expense_amount = float(input("Enter an expense: $"))
expense_description = input("What was this expense for? ")

# Categorise expense based on size
percent = (expense_amount / monthly_income) * 100

print("\n--- Expense Feedback ---")

if percent < 5:
    print("Category: Small Expense")
    print("Feedback: This is a minor expense. No problem.")

elif percent < 20:
    print("Category: Medium Expense")
    print("Feedback: This is a reasonable expense but keep track of similar spending.")

else:
    print("Category: Large Expense")
    print("Feedback: This is a large expense compared to your income. Consider budgeting carefully.")

# Show summary
print("\n--- Summary ---")
print(f"Expense: {expense_description}")
print(f"Amount: ${expense_amount:.2f}")
print(f"Percentage of income: {percent:.1f}%")
````
